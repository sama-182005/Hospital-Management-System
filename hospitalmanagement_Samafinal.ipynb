```python
class Person:
  def __init__(self,id,name,age,gender):
    self.id=id
    self.name=name
    self.age=int(age)
    self.gender=gender

  def __str__(self):
    return f" Name is {self.name},id={self.id}, Age is {self.age} , Gender is {self.gender} "

```

Patient Class
Attributes may include:
disease ,
visit_history

Responsibilities:
1.Validate patient-related data
2.Store patient information
3.Update patient details when required


```python
class Patient(Person):
  def __init__(self,id,name,age,gender,disease):
    super().__init__(id,name,age,gender)
    self.disease=disease
    self.visit_history=[]
    self._validate()

  def _validate(self):
    if self.age<=0:
      raise ValueError("Age cant be negative")
    if self.gender not in ["Male","Female","male","female"]:
      raise ValueError("Gender must be written as Male , Female")
    if not isinstance(self.name,str) or not self.name.strip():
      raise ValueError("Name must be a non_empty string")
    if not isinstance(self.disease,str) or not self.disease.strip():
      raise ValueError("Disease must be a non_empty string")

def update(self,update,value):
  update=update.lower()
  match update:
      case "name":
         self.name=value
      case "age":
        self.age=value
      case "disease":
        self.disease=value
      case _:
        raise ValueError("Invalid field to be updated")
  self._validate()

def Avisit(self,history):
  if not isinstance(history,str)or not history.strip():
    raise ValueError("History must be non_empty string")
  self.visit_history.append(history)

def __str__(self):
    return f"The patient whose id is {self.id} , his name is {self.name},age is {self.age} and his diagnosed disease is {self.disease}."
```

```python
class Doctor(Person):
  def __init__(self,id,name,age,gender,speciality):
    super().__init__(id,name,age,gender)
    self.speciality=speciality
    self.availability=True
    self.myschedule=set()
    self._validate()

  def _validate(self):
    if self.age<=0:
      raise ValueError("Age cant be negative")
    if self.gender not in ["Male","Female","male","female"]:
      raise ValueError("Gender must be written as Male , Female")
    if not isinstance(self.name,str) or not self.name.strip():
      raise ValueError("Name must be a non_empty string")
    if not isinstance(self.speciality,str) or not self.speciality.strip():
      raise ValueError("Disease must be a non_empty string")


  def update_availability(self,ava):
    if isinstance(ava,bool):
      self.availability=ava
    else:
      raise ValueError("Must enter True/False")

  def available_date(self,app_date):
    if self.availability:
      if not app_date in self.myschedule:
        self.myschedule.add(app_date)
      else:
        raise ValueError("This date is booked")



  def __str__(self):
    return f"The doctor whose id is {self.id},his name is {self.name},age is {self.age} , speciality is {self.speciality} and his availability is {self.availability}"
```

```python
from datetime import datetime
class Appointment:
  def __init__(self,app_id,patient_id,doctor_id,app_date,from_file=False):
    self.patient_id=patient_id
    self.doctor_id=doctor_id
    self.app_id=app_id
    self.app_date=app_date
    self.from_file=from_file
    self._validate()

  def _validate(self):
        if not self.app_id:
            raise ValueError("Appointment ID cant be empty.")

        if not self.patient_id:
            raise ValueError("Patient ID cant be empty.")

        if not self.doctor_id:
            raise ValueError("Doctor ID cant be empty.")


        try:
           self.app_date=datetime.strptime(self.app_date, "%Y-%m-%d").date()
        except ValueError:
            raise ValueError("Date must be in YYYY-MM-DD format.")


        if (not self.from_file) and (self.app_date<datetime.now().date()):
          raise ValueError("Date cant be in past")



  def __str__(self):
        return f"Appointment :{self.app_id} ,Patient: {self.patient_id},Doctor: {self.doctor_id},Date: {self.app_date}"
```

HospitalSystem Class

1.Manage collections of
patients, doctors, and appointments

2.Provide a menu-driven user interface

3.Coordinate operations between all classes

4.Handle saving and loading data using TXT files

5.Centralize exception handling to prevent system crashes

```python
import pandas as pd
```

```python

class Sys:
  def __init__(self):
    self.patients={}
    self.doctors={}
    self.Appointments={}

  def loading_patients(self,filename="patients.txt"):
    try:
      with open(filename,"r") as file:
        for line in file:
          patient_record=line.strip().split("|")

          if len(patient_record) !=5:
            print("This patient's record is not complete")
            continue

          #unpacking
          id,name,age,gender,disease=patient_record

          try:
            patient=Patient(id,name,age,gender,disease)
            #add this patient (object of Patient class) into patients dictionary
            #self.patients[id]=patient
            self.add_patient(patient)
          except ValueError as e:
            print(f"invalid data for: patient {id}:{e}")


    except FileNotFoundError:
      print("Patients file isnt found")

  def add_patient(self,patient):
      if(patient.id in self.patients):
        print("This patient is already registered")
      else:
        self.patients[patient.id]=patient

  def save_patients(self,filename="patients.txt"):
    try:
      with open(filename,"w") as file:
        for patient in self.patients.values():
          record=f"{patient.id}|{patient.name}|{patient.age}|{patient.gender}|{patient.disease}\n"
          file.write(record)
        print("Patients saved successfully.")
    except Exception as e:
      print(f"Error saving patients: {e}")

######################################################
#DOCTOR FUNCTIONS

  def loading_doctors(self,filename="doctors.txt"):
      try:
        with open(filename,"r") as file:
          for line in file:
            doctor_record=line.strip().split("|")

            if len(doctor_record) !=5:
              print("This doctor's record is not complete")
              continue

            id,name,age,gender,speciality=doctor_record
            try:
              d=Doctor(id,name,age,gender,speciality)
              self.add_doctor(d)
            except ValueError as e:
              print(f"Invalid data for doctor{id}:{e}")


      except FileNotFoundError:
        print("Doctors file not found.")

  def add_doctor(self,doctor):
    if doctor.id in self.doctors:
      print("This doctor is already registered")
      return
    self.doctors[doctor.id]=doctor


  def save_doctor(self,filename="doctors.txt"):
    try:
      with open(filename,"w") as file:
        for doctor in self.doctors.values():
          record=f"{doctor.id}|{doctor.name}|{doctor.age}|{doctor.gender}|{doctor.speciality}\n"
          file.write(record)
        print("Doctors Saved Succefully.")

    except Exception as e:
      print(f"Error saving doctors :{e}")


###########################################################
#Appointments
  def managing_appointments(self,app_id,p_id,d_id,a_date):
    if p_id not in self.patients:
      print("Patient is not found.")
      return
    if d_id not in self.doctors:
      print("Doctor is not found.")
      return

    doctor_obj=self.doctors[d_id]

    if doctor_obj.availability==False:
      print("Sorry, This doctor is not available.")
      return

    if a_date in doctor_obj.myschedule:
      print(f"Sorry, This date with {doctor_obj.name} is not available.")
      return



    try:
      app_obj=Appointment(app_id,p_id,d_id,a_date)
      if app_id in self.Appointments:
        print("This appointment is already booked.")
      else:
        """self.Appointments[app_id]=app
        doctor.myschedule.add(a_date)
        patient=self.patients[p_id]
        patient.visit_history.append(f"Appointment date :{app_id} with {doctor.name}")"""
        patient_obj=self.patients[p_id]
        self.appointment_helper(app_obj,doctor_obj,patient_obj)

        print("Appointment booked successfully.")
    except ValueError as e:
      print(f"Invalid data for appointment: {e} ")
      return



  def load_appointments(self,filename="appointments.txt"):
    try:
      with open(filename,"r") as file:
        for line in file:
          app_record=line.strip().split("|")
          if len(app_record) !=4:
            print("This appointment's record is not complete")
            continue

          app_id,p_id,d_id,a_date=app_record

          try:
            app_obj=Appointment(app_id,p_id,d_id,a_date,from_file=True)
            doctor_obj=self.doctors[d_id]
            patient_obj=self.patients[p_id]
            self.appointment_helper(app_obj,doctor_obj,patient_obj)
            """ self.Appointments[app_id]=app_obj
            doctor=self.doctors[d_id]
            doctor.myschedule.add(a_date)
            patient=self.patients[p_id]
            patient.visit_history.append(f"Appointment date :{app_id} with {doctor.name}")"""
          except ValueError as e:
              print(f"Invalid data for doctor{app_id}:{e}")

    except FileNotFoundError:
      print("Appointments file not found.")



  def save_appointments(self,filename="appointments.txt"):
    try:
      with open(filename,"w") as file:
        for app in self.Appointments.values():
          record=f"{app.app_id}|{app.patient_id}|{app.doctor_id}|{app.app_date}\n"
          file.write(record)
        print("Appointments Saved Succefully.")
    except Exception as e:
      print(f"Error saving appointment : {e}")


  def appointment_helper(self, app, doctor, patient):
    self.Appointments[app.app_id] = app
    doctor.myschedule.add(app.app_date)
    patient.visit_history.append(
        f"Appointment date: {app.app_date} with {doctor.name}"
    )

  def cancel_appointment(self, app_id):
    if app_id not in self.Appointments:
        print("Appointment not found.")
        return

    app = self.Appointments[app_id]
    doctor = self.doctors[app.doctor_id]
    patient = self.patients[app.patient_id]

    if doctor and app.app_date in doctor.myschedule:
        doctor.myschedule.remove(app.app_date)

    if patient:
       for v in patient.visit_history:
        if str(app.app_date) in v:
          patient.visit_history.remove(v)
          break



    del self.Appointments[app_id]

    print("Appointment cancelled successfully.")

#user methods
  def user_add_patient(self):
    try:
      p_id = input("Enter patient ID (or q to cancel): ").strip()
      if p_id.lower() == "q":
        print("Operation cancelled. Returning to menu.")
        return
      name = input("Enter patient name: ").strip()
      age = int(input("Enter patient age: "))
      gender = input("Enter patient gender: ").strip()
      disease = input("Enter patient disease: ").strip()

      patient_obj = Patient(p_id, name, age, gender, disease)
      self.add_patient(patient_obj)

      print("Patient added successfully.")

    except ValueError as e:
        print(f"Invalid input: {e}")

  def user_add_doctor(self):
    try:
      d_id=input("Enter doctor ID (or q to cancel): ").strip()
      if d_id.lower() == "q":
         print("Operation cancelled. Returning to menu.")
         return
      name=input("Enter doctor name: ").strip()
      age=int(input("Enter doctor's age: ").strip())
      gender=input("Enter doctor's gender: ").strip()
      speciality=input("Enter doctor's speciality:").strip()
      doctor_obj=Doctor(d_id,name,age,gender,speciality)
      self.add_doctor(doctor_obj)
      print("Doctor added successfully.")
    except ValueError as e:
      print(f"There is an invalid input: {e}")

  def user_book_appointment(self):
    try:
      app_id=input("Enter appointment ID (or q to cancel): ").strip()
      if app_id.lower() == "q":
         print("Operation cancelled. Returning to menu.")
         return
      p_id=input("Enter patient ID: ").strip()
      d_id=input("Enter doctor ID: ").strip()
      app_date=input("Enter appointment date: ").strip()
      self.managing_appointments(app_id,p_id,d_id,app_date)
    except Exception as e:
      print(f"There is an invalid input: {e}")



  def user_cancel_appointment(self):
    app_id = input("Enter appointment ID to cancel (or q to cancel): ").strip()
    if app_id.lower() == "q":
        print("Operation cancelled.")
        return

    self.cancel_appointment(app_id)


  def view_patients(self):
    if not self.patients:
      print("No patients found.")
      return
    patients_list=[]
    print("\n--- Patients List ---")
    for p in self.patients.values():
        patients_list.append({"ID": p.id,"Name": p.name,"Age": p.age, "Gender": p.gender,"Disease": p.disease, "Visit History": ", ".join(p.visit_history) if p.visit_history else "None"})
    df=pd.DataFrame(patients_list)
    display(df)

  def view_doctors(self):
    if not self.doctors:
      print("No doctors found.")
      return
    doctors_list=[] #list of dictionary
    print("\n--- Doctors List ---")
    for d in self.doctors.values():
      doctors_list.append({"ID": d.id,"Name": d.name,"Age": d.age, "Gender": d.gender,"Speciality": d.speciality})
    df=pd.DataFrame(doctors_list)
    display(df)

  def view_appointments(self):
    if not self.Appointments:
        print("No appointments found.")
        return

    df = pd.DataFrame([
        {
            "Appointment ID": a.app_id,
            "Patient ID": a.patient_id,
            "Doctor ID": a.doctor_id,
            "Date": a.app_date
        }
        for a in self.Appointments.values()
    ])

    display(df)




  def run(self):
    while True:
      print("\n===== Hospital Management System =====")
      print("1. Add Patient")
      print("2. Add Doctor")
      print("3. Schedule Appointment")
      print("4. View Patients")
      print("5. View Doctors")
      print("6. View Appointments")
      print("7. Cancel Appointment")
      print("8. Save & Exit")

      choice = input("Enter your choice: ").strip()
      match choice:
        case "1":
          self.user_add_patient()

        case "2":
          self.user_add_doctor()

        case "3":
          self.user_book_appointment()

        case "4":
          self.view_patients()

        case "5":
          self.view_doctors()

        case "6":
          self.view_appointments()

        case "7":
          self.user_cancel_appointment()

        case "8":
          self.save_patients()
          self.save_doctor()
          self.save_appointments()
          print("Data saved sucessfully!")
          break

        case _:
          print("Invalid choice. Try again.")
```

```python
S=Sys()
S.loading_patients()
S.loading_doctors()
S.load_appointments()
S.run()
```
